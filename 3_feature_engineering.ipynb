# 3. Feature engineering
Quitamos columnas que no sirven y convertimos las categóricas a números.

```python
import pandas as pd

df = pd.read_csv("data/titanic_clean.csv")
df.head()
```

## Columnas que no sirven
`PassengerId`, `Name` y `Ticket` son identificadores o texto único de cada pasajero y no tienen relación con que sobreviviera.

```python
df = df.drop(["PassengerId", "Name", "Ticket"], axis=1)
```

## Nueva variable: tamaño de la familia

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
```

## Codificación de variables categóricas
`Sex` y `Embarked` son texto; las convertimos en columnas de 0 y 1 (one-hot encoding).

```python
df = pd.get_dummies(df, columns=["Sex", "Embarked"], drop_first=True, dtype=int)
df.head()
```

```python
df.to_csv("data/titanic_fe.csv", index=False)
print("Datos listos guardados en data/titanic_fe.csv")
```
