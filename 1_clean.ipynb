# 1. Limpieza de datos
Cargamos el dataset del Titanic, revisamos valores nulos y decidimos qué hacer con cada columna.

```python
import pandas as pd

df = pd.read_csv("data/titanic.csv")
df.head()
```

```python
df.info()
```

## Porcentaje de valores nulos por columna

```python
porcentaje_nulos = (df.isnull().mean() * 100).round(2).sort_values(ascending=False)
porcentaje_nulos
```

## Columna `Cabin`
Tiene más del 75% de valores nulos, así que la eliminamos.

```python
df = df.drop("Cabin", axis=1)
```

## Columna `Age`
No eliminamos la edad porque es clave para el modelo. En el Titanic se salvó primero a mujeres y niños, así que la edad, junto con el sexo y la clase, influye mucho en quién sobrevivió. Rellenamos los faltantes con la mediana.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

## Columna `Embarked`
Solo le faltan 2 valores; los rellenamos con el puerto más común (moda).

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

## Verificación

```python
df.isnull().sum()
```

```python
df.to_csv("data/titanic_clean.csv", index=False)
print("Datos limpios guardados en data/titanic_clean.csv")
```
