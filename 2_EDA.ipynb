# 2. Análisis exploratorio (EDA)
Buscamos qué variables se relacionan con la supervivencia.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/titanic_clean.csv")
df.describe()
```

## ¿Cuántos sobrevivieron?

```python
sns.countplot(data=df, x="Survived")
plt.title("Sobrevivientes (1) vs no sobrevivientes (0)")
plt.show()

df["Survived"].value_counts(normalize=True).round(2)
```

## Supervivencia por sexo

```python
sns.barplot(data=df, x="Sex", y="Survived")
plt.title("Tasa de supervivencia por sexo")
plt.show()
```

## Supervivencia por clase

```python
sns.barplot(data=df, x="Pclass", y="Survived")
plt.title("Tasa de supervivencia por clase")
plt.show()
```

## Distribución de edad según supervivencia

```python
sns.histplot(data=df, x="Age", hue="Survived", bins=30, kde=True)
plt.title("Edad vs supervivencia")
plt.show()
```

## Correlación entre variables numéricas

```python
sns.heatmap(df.select_dtypes("number").corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Matriz de correlación")
plt.show()
```

**Conclusión:** las mujeres, los niños y los pasajeros de primera clase tuvieron más probabilidad de sobrevivir.
